```python
import openpyxl
path = r"student.xlsx" 
wb_obj = openpyxl.load_workbook(path)
sheet_obj = wb_obj.active
row = sheet_obj.max_row
column = sheet_obj.max_column

print("Total Rows:", row)
print("Total Columns:", column)

print("\nValue of first column")
for i in range(1, row + 1):
    cell_obj = sheet_obj.cell(row=i, column=1)
    print(cell_obj.value)

print("\nValue of second row")
for i in range(1, column + 1):
    cell_obj = sheet_obj.cell(row=2, column=i)
    print(cell_obj.value, end=" ")
```

```python
import openpyxl
from openpyxl.chart import BarChart, Reference
wb = openpyxl.Workbook()
sheet = wb.active

for i in range(10):
    sheet.append([i])
print("Chart Data:")
for row in sheet.iter_rows(min_row=1, max_row=10, min_col=1, max_col=1):
    print(row[0].value)

values = Reference(sheet, min_col=1, min_row=1, max_col=1, max_row=10)
chart = BarChart()
chart.add_data(values)
chart.title = "BAR-CHART"
chart.x_axis.title = "X_AXIS"
chart.y_axis.title = "Y_AXIS"
sheet.add_chart(chart, "E2")
wb.save("sample.xlsx")
```

```python
import openpyxl
from openpyxl.chart import LineChart, Reference

wb = openpyxl.Workbook()
sheet = wb.active

for i in range(10):
    sheet.append([i])  

print("Values for line chart:")
for row in sheet.iter_rows(min_row=1, max_row=10, min_col=1, max_col=1):
    print(row[0].value)


values = Reference(sheet, min_col=1, min_row=1, max_col=1, max_row=10)
chart = LineChart()
chart.add_data(values)
chart.title = "LINE-CHART"
chart.x_axis.title = "X-AXIS"
chart.y_axis.title = "Y-AXIS"
sheet.add_chart(chart, "E2")
wb.save("LINE.xlsx")
```

```python
import openpyxl
from openpyxl.styles import Font
wb = openpyxl.Workbook()
sheet = wb.active
sheet.cell(row=1, column=1).value = "I LOVE DOGS"
sheet.cell(row=1, column=1).font = Font(size=24)
sheet.cell(row=2, column=2).value = "I LOVE INDIA"
sheet.cell(row=2, column=2).font = Font(size=24, italic=True)
sheet.cell(row=3, column=3).value = "I LOVE INDIA"
sheet.cell(row=3, column=3).font = Font(size=24, bold=True)
sheet.cell(row=4, column=4).value = "I LOVE INDIA"
sheet.cell(row=4, column=4).font = Font(size=24, name='Times New Roman')
print("Styled Text in Excel:")
for row in [(1, 1), (2, 2), (3, 3), (4, 4)]:
    print(f"Row {row[0]}, Column {row[1]}: {sheet.cell(row=row[0], column=row[1]).value}")
wb.save('student.xlsx')
```

```python
import openpyxl
from openpyxl.chart import BarChart, Reference

# Create workbook and sheet
wb = openpyxl.Workbook()
sheet = wb.active

# Add data to the sheet
for i in range(10):
    sheet.append([i])

# Create reference to the data
values = Reference(sheet, min_col=1, min_row=1, max_col=1, max_row=10)

# Create the chart
chart = BarChart()
chart.add_data(values, titles_from_data=False)
chart.title = "BAR-CHART"
chart.x_axis.title = "X_AXIS"
chart.y_axis.title = "Y_AXIS"

# Add chart to the sheet
sheet.add_chart(chart, "E2")

# Save workbook
wb.save("student.xlsx")
```
